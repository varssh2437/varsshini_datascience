## Encoding Using Titanic Dataset

Encoding:
machine learning models works with only numbers

example:
sex
* Male
* Female
* A model cnnot directly cannot understand Male amd female so we convert categories into numbers using encoding

### Three types of encoding

    * label encoding 
    * one-hot encoding
    * target encoding

#### Label encoding
* assigns a number to each category
##### example:
size:
* small
* mediun
*large
##### becomes:
size:
* 0
* 1
* 2
###### the orde matters

#### when to use label encoding:
##### odinal categories:
###### example:
    * small<medium<large
    * low<medium<high
    * poor<average<Good

### Nominal categories
#### example:
* red,blue,green
* male,female
* India,USA,UK

### 2.One-pot Encoding(Nominal)
* creates seperate binary column for each category
##### when to use:
###### nominal categories
###### example:
    * country,gender,color,city
### 3.Target encoding
##### replace categories with the average value of the target variable.
###### when to use:
###### imagine: 5000 Cities
###### one-Hot encoding would create:
    * 5000 columns
    * not practical
##### taget encoding creates:
    * 1 Column


```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
#load Titanic dataset
df=sns.load_dataset('titanic')
#label Encoding
le=LabelEncoder()
df['sex_encoded']=le.fit_transform(
    df['sex']
)
# One-hot encoder
df=pd.get_dummies(
    df,
    columns=['embarked'],
    #drop_first=True
)
#Target Encoding
target_map=df.groupby('pclass')['survived'].mean()
df['class_taget']=(df['pclass'].map(target_map))
print(df.head())
```

### Outliers
* An outlier is a data point that is signficantly different from the rest of the data
##### why detect outlier:
##### outlier can :'
* disort the mean
* affect the deviation
* reduce model performance *influence scaling and regression mode
##### metod 1:visual detection using Box plot
*  A box plot quickly shows:
      * minimum value
      * Q1(25th quartile)
      * median
      * Q3(75th quartile)
      * maximum value
      * Outliers
#### method2: IQR Method:
* IQR=interquartile Range
* Formula=IQR=Q3-Q1
* steps:
  * Calculate IQR
  * calculate lower and upper limit
  * detect outlier
#### Method 3: Z-score method
* measures how many standard deviations a value is from the mean
* steps:
  * Step 1: Calculate Z-Score
  * step 2: Detect Outliers

```python
#method 1
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import zscore
# Load Titanic dataset
df=sns.load_dataset('titanic')
#boxplot
sns.boxplot(x=df['fare'])
plt.show()
# IQR Method
Q1=df['fare'].quantile(0.25)
Q3=df['fare'].quantile(0.75)
IQR=Q3-Q1
lower =Q1-1.5*IQR
upper=Q3+1.5*IQR
# varsshini value is equal to the infinity may be it can be less than zero
outliers_iqr=df[(df['fare']<lower) | (df['fare']>upper)]
print("IQR Outliers:",len(outliers_iqr))
# z_Score method
```

```python
# Method 2
# z_Score method
df['fare_zscore']=zscore(df['fare'])
outliers_z=df[abs(df['fare_zscore'])>3]
print("Z-score Outliers:",len(outliers_z))
```

### VIF (Variance Inflation Factor) for Multicollinearity Detection
#### what is Multicollinearity:
* Multicollinearty occurs when independent variables (features) are highly correlated with each other.
##### example:
    * suppose a house price dataset contains:
        * House Area(sq.ft)
        * no.of rooms
        *no of bedrooms
    * these features may contain similar information and be highly correlated.
    * this creates  multicollinearity.
#### What is VIF:
   * Variance inflation factor measures how much the variance of a feature is inflation because of its correlation with other features


| VIF Interpretation | Table |
| ------------------ | ----- |
| VIF                | interpretation |
|1                   | no correlation |
|1-5                 | low correlation |
| 5-10               |  moderate multi collinearity |
|>10                 | severe multicollinearity |

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
x=df.drop("species",axis=1)
vif=pd.DataFrame()
vif["feature"]=x.columns
vif["VIF"]=[
    variance_inflation_factor(x.values,i)
    for i in range(x.shape[1])
]
print(vif)
```

```python
# remove High-VIF Features
# suppose we remove petal_length.
x_new=x.drop("sepal_length",axis=1)
vif_after=pd.DataFrame()
vif_after["Feature"]=x_new.columns
vif_after["VIF"]=[
    variance_inflation_factor(x_new.values,i)
    for i in range(x_new.shape[1])
]
print(vif_after)
```
