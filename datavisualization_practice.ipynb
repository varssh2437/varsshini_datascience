```python
import numpy as np
import seaborn as sns
df=sns.load_dataset("tips")
print(df.head())
print(df.columns)
#2 descriptive statistics
print("Descriptive statistics")
print(df.describe())
#3.ientify skew distribution
print("skewness value")
numeric_cols=df.select_dtypes(include=np.number)
print(numeric_cols)
print(numeric_cols.skew())
#apply Z-score
zscore_df=numeric_cols.copy()
for col in zscore_df.columns:
    zscore_df[col]=(zscore_df[col]-zscore_df[col].mean())/zscore_df[col].std()
print(zscore_df.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
#5.pearson correlation
pearson_corr=numeric_cols.corr(method='pearson')
print(pearson_corr)
#heat map
plt.figure(figsize=(2,2))
sns.heatmap(
    pearson_corr,
    annot=True,
    cmap='coolwarm'
)
plt.title("pearson correlation")
plt.show()
#6.spearman_correlation
spearman_corr=numeric_cols.corr(method='spearman')
print(spearman_corr)
plt.figure(figsize=(2,2))
sns.heatmap(
    spearman_corr,
    annot=True,
    cmap='Reds'
)
plt.title("spearman correlation")
plt.show()
#7.compare pearson&spearman
comparison=pd.DataFrame({
    "pearson":pearson_corr["total_bill"],
    "spearman":spearman_corr["total_bill"]
})
print(comparison)
```
