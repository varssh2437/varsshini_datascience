 ## LINE CHART
 

```python
import matplotlib.pyplot as plt
months=['jan','feb','mar','apr']
sales=[100,150,200,250,]
plt.plot(months,sales,marker='o')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

## BAR CHART

```python
import matplotlib.pyplot as plt
dept=['HR','IT','Sales']
employees=[20,50,35]
plt.bar(dept,employees)
plt.title("Employees By Department")
plt.xlabel("Department")
plt.ylabel("count")
plt.show()
```

## PIE CHART

```python
import matplotlib.pyplot as plt
labels=['Food','Rent','Travel','others']
values=[40,50,60,20]
plt.pie(
    values,
    labels=labels,
    autopct='%1.1f%%'
    )
plt.title("Monthly Sales")
plt.show()
```

## SCATTER PLOT

```python
import matplotlib.pyplot as plt
hours=[2,3,4,5]
marks=[40,30,50,70]
plt.scatter(hours,marks)
plt.title("Student Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

## AREA CHART

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[10,15,30,10,50]
plt.fill_between(x,y)
plt.title("Area Chart")
plt.show()
```

## BOX CHART

```python
import matplotlib.pyplot as plt
salary=[25000,20000,30000,32000,35000,37000,40000,42000,45000,100000]
plt.boxplot(salary)
plt.title("Salary Distribution")
plt.ylabel("Salary")
plt.show()
```

## HEATMAP

```python
#correlation heatmap with seaborn
import seaborn as sns
import matplotlib.pyplot as plt
#Load dataset
tips=sns.load_dataset("tips")
#correlation matrix
corr=tips.corr(numeric_only=True)
#Heatmap
sns.heatmap(corr,annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

## HISTOGRAM

```python
#histogram using Seaborn
sns.histplot(
    data=tips,
    x="total_bill",
    bins=40
)
plt.title("Distribution of total Bill")
plt.show()
#sns.histplot()-creates a histogram
```

## KDE plot(Kernel Density Estimations)

```python
plt.figure(figsize=(9,3))
sns.kdeplot(
    data=tips,
    x="total_bill"
)
plt.title("KDE Plot")
plt.show()
```

```python
#Histogram +KDE Together
#Most commonly used in data science
plt.figure(figsize=(9,3))
sns.histplot(
    data=tips,
    x="total_bill",
    kde=True
)
plt.title("Histogram with KDE")
plt.show()
#kde=True ---> 
```

## SUBPLOTS

```python
#2 charts
import matplotlib.pyplot as plt
plt.figure(figsize=(8,4))
# First Plot
plt.subplot(1,2,2)
plt.plot([1,2,3], [10,20,30])
plt.title("Line Chart")
# Second Plot
plt.subplot(1,2,1)
plt.bar(['A','B','C'], [5,8,6])
plt.title("Bar Chart")
plt.show()
#Example: Four Charts
import matplotlib.pyplot as plt
plt.figure(figsize=(10,8))
plt.subplot(2,2,1)
plt.plot([1,2,3], [10,20,30])
plt.title("Line")
plt.subplot(2,2,2)
plt.bar(['A','B','C'], [5,8,6])
plt.title("Bar")
plt.subplot(2,2,3)
plt.scatter([1,2,3], [4,8,6])
plt.title("Scatter")
plt.subplot(2,2,4)
plt.pie([30,40,30], labels=['A','B','C'])
plt.title("Pie")
plt.tight_layout()
plt.show()

```
