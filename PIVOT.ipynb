```python
# creating a new column
#to_datetime()-converts the string date into actual data and time type
import pandas as pd
df=pd.read_csv("people-100.csv")
#step 2: Cretae age column
df["Date of birth"]=pd.to_datetime(df["Date of birth"])
today=pd.Timestamp.today()
df["Age"]=((today-df["Date of birth"]).dt.days//365)
print(df[["First Name","Sex","Age"]].head())
#2. Pivot table with aggfunc="mean"
pivot2=pd.pivot_table(
    df,
    values="Age",
    index="Sex",
    aggfunc="mean"
)
print(pivot2)
#1. Basic pivot table
# Average age by gender
pivot1=pd.pivot_table
df
values="Age"
index="Sex"
print(pivot1)
```

```python
# add overall average
pivot6=pd.pivot_table(
    df,
    values='Age',
    index='Sex',
    aggfunc='mean',
    margins=True
)
print(pivot6)
```
