```python
import pandas as p
a=p.Series([1,2,3,4,5])
print(a)
print(a.values)
print(a.index)
print(a.dtype)
g=p.Series([True,False,True,True,])
print(g.dtype)
#Accessing Elements
print(a[1])
#Accessing Elements using loc
print(a.iloc[1])
#Slicing
d=p.Series(['devasri','monika','kavya','tulsi','vish','sanjeevi','samicsha'])
print(d[1:3])          # stop value is exclusive
print(d.loc[1:3])      # stop value is inclusive
print(d.iloc[2:5])     # stop value exclusive
```

```python
#setting index value
#we can define our own index
import pandas as p
d1=p.Series(['kavya','tulsi','devasri','monika','vish','sanjeevi','samicsha'])
print(d1)
d2=p.Series(['kavya','tulsi','vish','devasri','monika','sanjeevi','samicsha'],index=['a','b','c','d','e','f','g']) #labled indexs
print(d2['e'])
print(d2['a']+'vish')                                                                                              #vectorized operation - no loop required for this

'''
in iloc we cannot be use labled index
iloc can be used with inly the original index values
using loc elements can be accessed with the labled indexs
'''
print(d1.iloc[1])
print(d2.iloc[1])
print(d2.loc['a'])
# Statistical Function
d3=p.Series([11,2,3,45,-5,6,7,8])
print(d3.sum())
print(d3.max())
print(d3.abs())
print(d3.min())
print(d3.mean())
print(d3.median())        # Cannot use percentile Function
print(d3[d3>5])           # Boolean Functions
print(len(d3))            # count values in array
print(d3.value_counts())  #returns individual value count
#String Operations
print(d1.str.upper())
print(d1.str.len())
```

```python
# Opening json file:
import pandas as p
df=p.read_json("titanic.json")
df.head()
#check null values- isnull() - use to identify missing values
print(df.isnull())
# count missing values for all field
print(df.isnull().sum())
#Find rows having missing age
df[df['Age'].isnull()]
#to view what column are there
print(df.columns)
#replace missing values-fillna()
mean_age=df['Age'].mean()
df['Age']=df['Age'].fillna(mean_age)
print(df.isnull().sum())
#Fill Embarked with most frequenr value
#by using mode() means the most frequently occuring value in a column
mode_embarked=df['Embarked'].mode()[0]
df['Embarked']=df['Embarked'].fillna(mode_embarked)
print(df.isnull().sum())
print(df)
#dropna()
#remove missing data
#remove Rows
df.dropna()
#remove Columns
df.dropna(axis=1)
#groupby
#Average fare per class
print(df.groupby('Pclass')['Fare'].mean())
#count Survivors By Gender
df.groupby('Sex')['Survived'].sum()
#multiple Aggregations-agg()
print(df.groupby('Pclass').agg({
    'Age':['mean','min','max'],
    'Fare':['mean','sum']}))
#transform()
#returns same number of rows
#add Class Average Fare
df['Avg_class_fare']=df.groupby('Pclass')['Fare'].transform('mean')
print(df.to_string())
```

## MERGING SHEETS INTO ONE SHEET IN EXCEL FILE

```python
import pandas as p
students=p.read_excel("Merge_join_Concat_Dataset_50_Records.xlsx",sheet_name="students")
marks=p.read_excel("Merge_join_Concat_Dataset_50_Records.xlsx",sheet_name="marks")
contacts=p.read_excel("Merge_join_Concat_Dataset_50_Records.xlsx",sheet_name="contact")
## 1. INNER MERGE
#Returns only matching Students_IDs from both Data Frames
inner_df=p.merge(students,marks,on="Student_ID",how="inner")
print(inner_df.head())
print(inner_df.shape)
#LEFT MERGE
left_df=p.merge(students,marks,on="Student_ID",how="left")
print(left_df.tail())
print(left_df.shape)
#even if marks missing,student remain.
#RIGHT MERGE
right_df=p.merge(students,marks,on="Student_ID",how="right")
print(right_df.tail())
print(right_df.shape)
#OUTER MERGE
outer_df=p.merge(students,marks,on="Student_ID",how="outer")
print(outer_df.tail())
print(outer_df.shape)
```

## MERGE MULTIPLE TABLES

```python
#Combine student+marks+cotacts
import pandas as p
students=p.read_excel("Merge_join_Concat_Dataset_50_Records.xlsx",sheet_name="students")
marks=p.read_excel("Merge_join_Concat_Dataset_50_Records.xlsx",sheet_name="marks")
contacts=p.read_excel("Merge_join_Concat_Dataset_50_Records.xlsx",sheet_name="contact")
student_complete=p.merge(students,marks,on="Student_ID",how="left")
student_complete=p.merge(student_complete,contacts,on="Student_ID",how="left")
print(student_complete.head())
```

## JOIN()
### merge-worksbased on the id(merging two tables without using common column)
### join-works based on indexvalue (join two tabels using a common column)

```python
import pandas as p
students=p.read_excel("Merge_join_Concat_Dataset_50_Records.xlsx",sheet_name="students")
marks=p.read_excel("Merge_join_Concat_Dataset_50_Records.xlsx",sheet_name="marks")
contacts=p.read_excel("Merge_join_Concat_Dataset_50_Records.xlsx",sheet_name="contact")
students_index=students.set_index("Student_ID")
contact_index=contacts.set_index("Student_ID")
join=students_index.join(contact_index)
print(join.tail(10))
```
