## Encoding Using Titanic Dataset

Encoding:
machine learning models works with only numbers

example:
sex
* Male
* Female
* A model cnnot directly cannot understand Male amd female so we convert categories into numbers using encoding

### Three types of encoding

    * label encoding 
    * 

#### Label encoding
* assigns a number to each category
##### example:
size:
* small
* mediun
*large
##### becomes:
size:
* 0
* 1
* 2
###### the orde matters

#### when to use label encoding:
##### odinal categories:
###### example:
    * small<medium<large
    * low<medium<high
    * poor<average<Good

### Nominal categories
#### example:
* red,blue,green
* male,female
* India,USA,UK

### 2.One-pot Encoding(Nominal)
* creates seperate binary column for each category
##### when to use:
###### nominal categories
###### example:
    * country,gender,color,city
### 3.Target encoding
##### replace categories with the average value of the target variable.
###### when to use:
###### imagine: 5000 Cities
###### one-Hot encoding would create:
    * 5000 columns
    * not practical
##### taget encoding creates:
    * 1 Column


```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
#load Titanic dataset
df=sns.load_dataset('titanic')
#label Encoding
le=LabelEncoder()
df['sex_encoded']=le.fit_transform(
    df['sex']
)
# One-hot encoder
df=pd.get_dummies(
    df,
    columns=['embarked'],
    #drop_first=True
)
#Target Encoding
target_map=df.groupby('pclass')['survived'].mean()
df['class_taget']=(df['pclass'].map(target_map))
print(df.head())
```

### Feature Scaling Using Titanicataset
#### What is feature scaling
##### Feature scaling transforms numerical feature to similar scale
#### when it is needed
###### In titanic dataset
###### Example : feature Range Age0 -80 fare 0-512
###### Here ,Fare has a much larger range than age.
###### Many ML algorithm calculates distance,so Fare can dominate the model.

#### 1.MinMAxScalar
##### scales values between 0 and 1.
##### Advantages:
 * preserves shape of distribution
 * values always between 0 and 1
 * useful for nerual networks
#### Disadvantages:
 * Sensitive to outliers
#### 2.standardScalar
 centers data around mean =0 and standard deviation=1
 Advantages:
  * commonly well for normally distributed data
  * commonly used
  * required by many algorithms
#### Best for:
##### logistic regreeion ,SVM,PCA,KNN,Linear Regression,
#### Disadvantages:
  * Sitll affected by extreme outliers
#### 3.RobustScaler
##### Uses Median and IQR insteard of mean and standard deviation
##### Advantages:
 * handles outliers well
 * better for skewed distributions
##### Disadvantages:
 * Values are not limited to 0-1

```python
from sklearn.preprocessing import MinMaxScaler
scalar=MinMaxScaler()
df[['age_minmax','fare_minmax']]=scalar.fit_transform(
    df[['age','fare']])
df[['age_minmax','fare_minmax']].head()
```

```python
import matplotlib.pyplot as plt

plt.hist(df['fare'].dropna(),bins=30)
plt.title("Original Fare")
plt.show()
plt.hist(df['fare'],bins=30)
plt.title("MinMax Scaled Fare")
plt.show()
```
