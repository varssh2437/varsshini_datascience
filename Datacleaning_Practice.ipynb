```python

import pandas as pd
df=pd.read_csv("Messy_Employee_dataset.csv")
print(df)
print(df.columns)
print(df.isnull().sum())
mean_age=df['Age'].mean()
df['Age']=df['Age'].fillna(mean_age)
print(df['Age'])
print(df.isnull().sum())
df['Salary']=df['Salary'].fillna(mean_age)
print(df['Salary'])
print(df.isnull().sum())
duplicates=df.duplicated()
print(duplicates)
```
