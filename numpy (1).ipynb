```python
#array creation in numpy
'''
1 np.array() - converts list or tuple into array
2 np.zeros() - creating a array with initial value zero
3 np.ones() -  creating a array with initial value one
4 np.arange() - works like range() function
    syntax  
    np.arange(start,stop,step)
'''
import numpy as n
a=n.array([1,3,2,4,5])
b=n.zeros(5)
print(a)
print(b)
c=n.zeros((2,2))
print(c)
d=n.ones(2)
print(d)
e=n.ones((2,4))
print(e)
print(n.arange(1,10,2))                 # splits elements into 2 intervals
print(n.linspace(1,20,5))               # splits into 5 equal elements
print(n.random.rand(5))                 # print values form 0 to 1
print(n.random.randint(-100,100,20))    # print values with the range and no of elements specified in right side
print(n.random.randn(5))                # prints random decimal,can be positive and negative
```

```python
#Indexing&slicing
'''
used for
1 accessing data
2 extract data
3 modify data
numpy supports
1 1D-indexing
2 2D-indexing
3 1D-slicing
4 2D-slicing
5 Boolean indexing
6 Fancy indexing
slicing syntax
array_name(start:stop:step)
'''
import numpy as n
a=n.array([10,20,30,40,50])
print(a[1])
print(a[2:5])
print(a[:4])
print(a[:-1])
b=n.array([[1,2,3],[2,5,6],[4,7,8]])
print(b[1][2])
print(b[1])
print(b[:,-1])
```

```python
#Boolean Indexing -use to filter values based on conditions
#Fancy Indexing -use to print specific multipe values from the array
import numpy as n
a=n.array([10,20,30,40,50])
print(a[a>25])             # Boolean Indexing
print(a[a%2==0])           # Boolean Indexing
print(a[[0,4,2,3]])        # Fancy Indexing
a[3]=10
print(a)
```

```python
# Mathematical Operation in numpy
'''
1 Element wise operation
2 Broadcasting - it allows numpy to perform operation between array of different shapes
Types:
    1-Scalar Broadcasting
    2-2D Broadcasting
    3-Row Broadcasting
        rule:two array shoud have same dimensions,which means both array should have same no of rows 
3 Matrix Operation - 
'''
import numpy as n
a=n.array([10,20,30])
b=n.array([[1,2,3],[2,3,4]])
print(a+5)                   # Scalar Broadcasting
print(b+10)                  # 2D  Broadcasting
print(b+a)                   # Row Broadcasting
# Matrix Oerations
c1=n.array([[1,2],[3,4]])
c2=n.array([[2,3],[8,9]])
print(a*b)
print(np.matmul(c1,c2))      # or print(np.dot(c1,c2)) or print(a@b)
print(a.T)                   #Transpose(row into colunm & viceversa)
```

```python
#Statistical Functions
'''
used to:
    1 Data Analysing
    2 to find average
    3 Spread of data
    4 find relationship between variable
1 Mean()-find average
2 Median()-to find mid value afer sorting
3 std deviation()-how far data spread from mean
    small std-data close to mean
    large std-data spread widely
4 coorelation-relationship between variable
    positve= +1
    negative= -1
    no coorelation
5.Persentile
'''
import numpy as n
a=n.array([10,2,3,30,50])
print(n.mean(a))
print(n.median(a))
print(n.std(a))
print(n.min(a))
print(n.max(a))
print(n.percentile(a,25))
print(n.var(a))
print(n.argmin(a))
print(n.argmax(a))
b=n.array([10,2,3,30,50])
c=n.array([20,12,13,40,60])
print(n.corrcoef(b,c))

# Aggregate Function
d=n.array([[2,3,4],[2,4,5]])
print(n.sum(d))
print(n.sum(d,axis=1))
print(n.sum(d,axis=0))
```
